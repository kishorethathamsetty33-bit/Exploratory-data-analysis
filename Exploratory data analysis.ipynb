```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Kishore\Downloads\netflix_titles.csv",encoding='latin-1')
df
```

```python
df.info()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["director"].fillna(df['director'].mode()[0],inplace=True)
```

```python
df["cast"].fillna(df['cast'].mode()[0],inplace=True)
```

```python
df["country"].fillna(df['country'].mode()[0],inplace=True)
```

```python
df["date_added"].fillna(df['date_added'].mode()[0],inplace=True)
```

```python
df.loc[:,"rating"]=df['rating'].fillna(df['rating'].mode()[0])
```

```python
df.loc[:,"duration"]=df['duration'].fillna(df['duration'].mode()[0])
```

```python
df=df.dropna(axis=1,how='all')
```

```python
df.isnull().sum()
```

```python
df
```

```python
#distribution of release year
sns.histplot(df['release_year'])
plt.title("distribution of release year")
plt.show()
```

```python
#rating
sns.countplot(y='rating',data=df,order=df['rating'].value_counts().index)
plt.title("Rating Distribution")
plt.show()
```

```python
sns.countplot(x='type',data=df)
plt.title("Movies vs Tv Shows")
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 countries by content")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```

```python
df['listed_in'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.show()
```
